### Based on conditional statements (if, if-else, if-elif-else) and nested if-else

```python
# 1.       Write a program to assign grades based on marks:
# -          Marks ≥ 90: Grade A
# -          Marks ≥ 75: Grade B
# -          Marks ≥ 50: Grade C
# -          Else: Fail

marks = 92;

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

```python
#2. Write a program to display the day of the week based on a number (1 for Monday, 2 for Tuesday, etc.).
dayNum = int(input("Enter a number from 1-7 : "))

if dayNum == 1:
    print("Its Monday..")
elif dayNum == 2:
    print("Its Tuesday..")
elif dayNum == 3:
    print("Its Wednesday..")
elif dayNum == 4:
    print("Its Thrusday..")
elif dayNum == 5:
    print("Its Friday..")
elif dayNum == 6:
    print("Its Saturday..")
elif dayNum == 7:
    print("Its Sunday..")
else:
    print("Enter a valid number!")
```

```python
# 3. Write a program to calculate electricity bill based on units consumed:
# - Up to 100 units: ₹5/unit
# - 101–200 units: ₹7/unit
# - Above 200 units: ₹10/unit

UnitsConsumed = 120

if UnitsConsumed <= 100:
    print(f'Total Cost = Rs{UnitsConsumed*5}')
elif UnitsConsumed <= 200:
    print(f'Total Cost = Rs{UnitsConsumed*7}')
else:
    print(f'Total Cost = Rs{UnitsConsumed*10}')
```

```python
# 4.Write a program to check whether a number is positive and even or positive and odd and also negative even or negative odd

WholeNum = 5.6

if WholeNum%2 == 0 and WholeNum > 0:
    print("The Number = ",WholeNum, "is Even Positive")
elif WholeNum%2 !=0 and WholeNum > 0:
    print("The Number = ",WholeNum, "is Odd Positive")
elif WholeNum%2 == 0 and WholeNum < 0:
    print("The Number = ",WholeNum, "is Even Negative")
elif WholeNum%2 != 0 and WholeNum < 0:
    print("The Number = ",WholeNum, "is Odd Negative")
else:
    print("Enter a valid whole number")
```
