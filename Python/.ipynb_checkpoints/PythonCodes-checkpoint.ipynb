### Looping Statements:

#### While Loop 

```python
# Write a program to print numbers from 1 to 10 using while loop.

count = 1

while (count <= 10):
    print("The count is",count)
    count = count + 1

print("Out of while loop")
```

```python
# Write a program to print numbers from 10 to 1 using while loop.

count = 10

while (count >= 1):
    print("The count is",count)
    count = count - 1

print("Out of while loop")
```

```python
nu = 563
tot = 0

digit = num % 10 # 563 % 10 = 3
print("DIGIT",digit) # 3
tot = tot + digit # 0 + 3 = 3
print("TOTAL",tot) # 3
nu = nu // 10 # 563 // 10 = 56
print("Number",nu) # 56

digit = nu % 10  # 56 % 10 = 6
print("DIGIT",digit) # 6
tot = tot + digit # 3 + 6 = 9
print("Total",tot) # 9
nu = nu // 10  # 56 // 10 = 5

digit = nu % 10  # 5 % 10 = 5
print("DIGIT",digit) # 5
tot = tot + digit # 9 + 5 = 14
print("Total",tot) # 14
nu = nu // 10  # 0
```

### Nested For Loops

```python
# repeat hore hai number
num = 100
i = 1

while i <= num:
    if i%3 == 0 and i%5 == 0:
        print(i,"FizzBuzz")
    elif i%3 == 0:
        print(i,"Fizz")  
    elif i%5 == 0:
        print(i,"Buzz")
    else:
        print(i)
    i = i + 1
```

```python
tot = 0
for i in range(5):
    num = int(input("Enter any number = "))    # multiple time user se input lene ke liye
    if (num == 0):
        break
    else:
        tot = tot + num

print("Total = ",tot)
```

```python
t = 0

while True: # naya logic hai --> hamesha 1 dega
    n = int(input("Enter num : "))
    if n == 0:
        break
    t = t + n

print("Total = ",t)
```

```python
#3. Login System : Allow a user a maximum of 3 Login attempts.
# If the correct password is entered, stop the loop .
# Try after 24 hours.

SignUp_UserName = input("Enter your new username = ")
SignUp_Password = input("Set your account password = ")
```

```python
SignIn_UserName = input("Enter your username = ")
SignIn_Password = input("Enter your password = ")

for i in range(4):
    if SignUp_UserName != SignIn_UserName or SignUp_Password != SignIn_Password:
        print("Invalid username or password")
        SignIn_UserName = input("Enter your username = ")
        SignIn_Password = input("Enter your password = ")
        continue
        
    elif SignUp_UserName == SignIn_UserName and SignUp_Password == SignIn_Password:
        print("Login-Successfull")
        break

else :
    print("Try After 24 hours")
```

```python
# Check weather the given number is prime or not.

num = int(input("Enter a number = "))

i = 1
while i <= num:
    if num%i == 0 and num/1 == 1:
        print(num,"is a prime number")
        break
    else:
        print(num,"is not a prime number")
    i += 1
```
