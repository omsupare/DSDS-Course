### Control Flow in Python

#### Conditional Statements :


```python
Age = int(input("Enter your age: "))

if Age >= 18:
    print("You are elligible to vote")
    print("Thank You")

print("Successfully Logged Out")
```

```python
VoterAge = int(input("Enter your age : "))
VoterCity = input("From which city do u belong ?")

if VoterAge >= 18 and VoterCity == "Nagpur":
    print("You are elligible to vote")
    print("Thank You")

print("Successfully Logged Out")
```

##### NestedIf

```python
VoterAge = int(input("Enter your age : "))
VoterCity = input("From which city do u belong ?")

if VoterAge >= 18:
    if VoterCity == "Nagpur":
        print("You are elligible to vote")
        print("Thank You")

print("Successfully Logged Out")
```

#### if-else statement

```python
marks = int(input("Enter your marks = "))

if marks >= 40:
    print("You have passed your exam")
    print("Congratulations")
else:
    print("You have failed your exam")
```

### Nested if-else

```python
n1 = int(input("Enter first number"))
n2 = int(input("Enter second number"))

if n1 >= n2:
    if n1 == n2:
        print(f'{n1} is equal to {n2}')
    else:
        print(f'{n1} is greater than {n2}')
else:
    print(f'{n1} is less than {n2}')
```

### Check multiple conditions using if-elif

```python
marks = int(input("Enter your marks : "))   # ager if lagaya toh voh sare step by step execute hoga pr elif me false hone pr hi hoga aaise

if marks >= 75:
    print(f'Grade: A+')
elif marks >= 50:
    print(f'Grade: A')
elif marks >= 40:
    print(f'Grade: B+')
else:
    print(f'Failed')
```
